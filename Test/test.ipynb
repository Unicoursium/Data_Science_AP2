```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1  读取原始 CSV（记得分号分隔）
df = pd.read_csv('marketing_campaign_CLEANED.csv', sep=',')

# 2  计算非红酒总消费
df['TotalSpent'] = df[['MntFruits','MntMeatProducts','MntFishProducts',
                       'MntSweetProducts','MntGoldProds']].sum(axis=1)

print(df['MntWines'].quantile([.75,.8,.9,.95]))

# 3  画直方图 + KDE
plt.figure(figsize=(6,4))
sns.histplot(df['MntWines'], bins=40, kde=True)
plt.xlabel('TotalSpent (€) - non-wine categories (2 yrs)')
plt.title('Distribution of TotalSpent')
plt.tight_layout()
plt.show()
```
